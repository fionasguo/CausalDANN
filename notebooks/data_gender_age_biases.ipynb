```python
import pandas as pd
import numpy as np
import re
```

```python
def get_gender(text):
  try:
    marker = text.split('I (')[1].split(')')[0]
    clean = re.sub(r'[0-9,./-]', '', marker).lower()
    female = np.any([t == 'f' or t == 'female' for t in clean.split(' ')])
    male = np.any([t == 'm' or t == 'male' for t in clean.split(' ')])
    return +1 if female else (-1 if male else 0)
  except:
    return 0

def get_age(text):
  try:
    marker = text.split('I (')[1].split(')')[0]
    clean = re.sub(r'[a-z,./ -]', '', marker.lower())
    return int(clean)
  except:
    return -1
```

```python
def get_gender(text: str) -> str:
    """
    Detects self-claimed gender from Reddit posts.
    Returns 'male', 'female', or 'undetermined'.
    """
    try:
        marker = text.split('I (')[1].split(')')[0]
        clean = re.sub(r'[0-9,./-]', '', marker).lower()
        female = np.any([t == 'f' or t == 'female' for t in clean.split(' ')])
        male = np.any([t == 'm' or t == 'male' for t in clean.split(' ')])
        if female:
            return +1
        elif male:
            return -1
        # return +1 if female else (-1 if male else 0)
    except:
        pass
    text = str(text).lower()
    
    female_patterns = [
        r"i am a female", r"i'm a female", r"i am a woman", r"i'm a woman",
        r"as a female", r"as a woman", r"i am a girl", r"i'm a girl",
        r"i \\(f\\)", r"me \\(f\\)", r"\(f\)", r"\[f\]", r"\{f\}"
    ]
    
    male_patterns = [
        r"i am a male", r"i'm a male", r"i am a man", r"i'm a man",
        r"as a male", r"as a man", r"i am a boy", r"i'm a boy",
        r"i \\(m\\)", r"me \\(m\\)", r"\(m\)", r"\[m\]", r"\{m\}"
    ]
    
    for pattern in female_patterns:
        if re.search(pattern, text):
            return +1 # female
    
    for pattern in male_patterns:
        if re.search(pattern, text):
            return -1 # "male"
    
    return 0 # "undetermined"
```
